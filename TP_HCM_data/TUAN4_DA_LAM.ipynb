# B. Ý 4 — Ceteris paribus ⭐ phần quan trọng nhất

**File:** `analysis/14_ceteris_paribus.ipynb` · hình `CP1`–`CP9`

Đo bằng **hai đường độc lập**: hỏi model (partial dependence) và hỏi dữ liệu thật (đối chứng
ghép cặp). Hai đường khớp ⇒ tin được; lệch ⇒ biết model học nhầm chỗ nào.

## B1. Phát hiện lớn nhất — model dự đoán giỏi nhưng không hiểu cơ chế

| Hỏi model | Model trả lời | Dữ liệu thật | Lệch |
|---|---:|---:|---:|
| Trời chuyển mưa, giá đổi bao nhiêu? | **+0,93%** | **+11,05%** | −10,1 điểm |
| Giờ cao điểm so với giờ thường? | **+2,42%** | **+12,82%** | −10,4 điểm |

Lệch **cùng hướng, cùng cỡ** ⇒ triệu chứng cấu trúc, không phải nhiễu.

**Nguyên nhân với giờ cao điểm:** model dùng `latest_observed_*` và `history_60m_*` — giá đối thủ
vừa quan sát **đã chứa sẵn** tác động của giờ. Bỏ nhóm feature đó rồi train lại → **+13,4%**,
khớp thực tế (`CP7b`).

## B2. Cơ chế đầy đủ của mưa — truy được 83%

Mưa thì bỏ feature không cứu được (+0,83%). Nên phải đào tiếp. Mở dần từng đường bị chặn (`CP9`):

| Bước | Mưa đẩy giá | Giải thích được |
|---|---:|---:|
| Model đầy đủ, chặn mọi đường | +0,93% | 0% |
| + mở cung–cầu, tỷ lệ toàn cục | +3,01% | 21% |
| + tỷ lệ riêng từng ô | +3,00% | 21% |
| + bỏ feature giá quan sát | +5,60% | 46% |
| **+ mở đường THỜI LƯỢNG** | **+9,37%** | **83%** |
| Dữ liệu thật | +11,05% | 100% |

```
                    ┌── tắc đường ──→ thời lượng +13,1% ──→ giá cơ bản ↑   (37 điểm)
   TRỜI MƯA ────────┤
                    └── cầu +15,5% ──→ mất cân bằng +24,8% ──→ hệ số nhân ↑ (21 điểm)
                        cung −7,0%
                                                    tổng cộng: +11,05%
```

**Mưa không đẩy giá trực tiếp một chút nào.** Toàn bộ đi qua hai đường trung gian, đường tắc
đường mạnh gần gấp đôi đường cung–cầu.

### Ba nghi ngờ ban đầu, kiểm hết

| Nghi ngờ | Kết quả |
|---|---|
| Tỷ lệ cung–cầu trung bình toàn cục quá thô | ❌ **Sai** — dùng tỷ lệ riêng từng ô: 3,01% → 3,00% |
| Đối chứng chưa khống chế theo ngày | ❌ **Sai** — thêm ngày: hiệu ứng vẫn 9,82–11,05% |
| Mưa còn đi qua thời lượng | ✅ **Đúng, đường lớn nhất** |

> **Kiểm chứng khoá đối chứng có chạy đúng không:** trong cùng ô, `quote_distance` giữa nhóm mưa

> Nếu quãng đường cũng lệch thì mọi kết luận trên đã vô nghĩa.

## B4. Đối chứng ghép cặp — cách làm

Chia test thành **ô giống nhau**: loại xe × tuyến × khung km 1 dặm × giờ. Trong mỗi ô so giá thật
giữa hai nhóm, trung bình có trọng số theo số chuyến. Chỉ giữ ô có **cả hai nhóm**, mỗi nhóm
**≥ 30 chuyến**. Đã bỏ trùng lag (mỗi chuyến đếm 1 lần, không phải 4).

| Hiệu ứng | Chênh giá | KTC 95% | t | Đồng đều giữa các ô? |
|---|---:|---|---:|---|
| Mưa | **+11,05%** | [+10,52%, +11,59%] | 40 | đều |
| Cao điểm | **+12,82%** | [+12,27%, +13,37%] | 46 | đều |
| Cuối tuần | **+5,98%** | [+5,19%, +6,77%] | 15 | ⚠️ **không** (sd 11,1%) |

> **Cách đọc:** `sd` (phân tán giữa các ô) và `se` (sai số của trung bình) là **hai thứ khác nhau**.
> Cuối tuần có sd lớn nhưng 759 ô nên se chỉ 0,40 điểm — hiệu ứng vẫn chắc chắn có, chỉ là không đều.

# E. Kiểm tự động

````python
import json, sys, warnings
warnings.filterwarnings("ignore")
import numpy as np, pandas as pd
from pathlib import Path

R = Path(".")
H = R / "docs" / "hinh_anh"
loi = 0

print("=" * 78)
print("1. NOTEBOOK — chay het chua, co loi khong")
print("=" * 78)
NB = ["analysis/14_ceteris_paribus.ipynb",
      "model/uncertainty/00_TONG_QUAN.ipynb",
      "model/uncertainty/01_conformal_chuan_hoa.ipynb",
      "model/uncertainty/02_quantile_regression.ipynb",
      "model/uncertainty/03_CQR.ipynb",
      "model/uncertainty/04_SO_SANH.ipynb",
      "model/uncertainty/05_PHAN_RA_theo_thoi_gian.ipynb",
      "model/train/07_sinh_du_lieu_UQ.ipynb",
      "model/evaluation/08_truc_quan_GAM.ipynb",
      "model/train/05_GAM_transformed_feature_space.ipynb"]
for f in NB:
    p = R / f
    if not p.exists():
        print(f"  [!!!] {f} — KHONG TON TAI"); loi += 1; continue
    nb = json.load(open(p, encoding="utf-8"))
    cod = [c for c in nb["cells"] if c["cell_type"] == "code"]
    chua = sum(1 for c in cod if not c.get("execution_count"))
    e = sum(1 for c in cod for o in c.get("outputs", [])
            if o.get("output_type") == "error")
    loi += chua + e
    print(f"  [{'OK ' if chua == 0 and e == 0 else '!!!'}] {f:50s} "
          f"{len(cod):2d} cell · chua chay {chua} · loi {e}")

print()
print("=" * 78)
print("2. HINH — co du khong")
print("=" * 78)
co = {p.name for p in H.glob("*.png")}
for nhom, n, ten in [("CP", 9, "ceteris paribus"), ("BM", 3, "band 3pp x 3muc"),
                     ("QR", 4, "quantile regression"),
                     ("CQ", 4, "CQR"),
                     ("TT", 4, "sai so theo thoi gian"), ("BG", 4, "band gia"),
                     ("GA", 6, "GAM"), ("TF", 4, "tuong tac GAM")]:
    thay = sorted(p for p in co if p.startswith(nhom) and p[2] in "0123456789")
    tt = "OK " if len(thay) >= n else "!!!"
    if len(thay) < n:
        loi += 1
    print(f"  [{tt}] {nhom}* ({ten:22s}): {len(thay)}/{n} hinh")

print()
print("=" * 78)
print("3. SO LIEU — tinh lai tu du lieu, doi chieu bang o tren")
print("=" * 78)
E = R / "model" / "evaluation"
ut = pd.read_parquet(E / "uq_pred_test.parquet")
uc = pd.read_parquet(E / "uq_pred_calibration.parquet")
q90 = np.quantile(((uc.hybrid_pred - uc.gia_that).abs() / uc.hybrid_pred).values, .90)

b = ut[ut.hybrid_pred >= 300e3]
cov = ((b.gia_that >= b.hybrid_pred*(1-q90)) &
       (b.gia_that <= b.hybrid_pred*(1+q90))).mean()
print(f"  Band >300k coverage : {cov:.2%}   (bang tren: 82,7%)")

u2 = ut.assign(cd=ut.gio_vn.apply(lambda h: (7 <= h <= 9) or (17 <= h <= 19)),
               sai=lambda d: (d.hybrid_pred - d.gia_that).abs() / d.gia_that,
               ae=lambda d: (d.hybrid_pred - d.gia_that).abs())
g = u2.groupby("cd").agg(sai=("sai", "mean"), mae=("ae", "mean"))
print(f"  MAPE cao diem       : {g.sai[True]:.2%} · gio thuong {g.sai[False]:.2%} "
      f"-> chenh {abs(g.sai[True]-g.sai[False])*100:.3f} diem (kich ban A)")
gio = u2.groupby("gio_vn").ae.mean()
print(f"  MAE theo gio        : {gio.min():,.0f}d (gio {gio.idxmin()}) -> "
      f"{gio.max():,.0f}d (gio {gio.idxmax()}) = +{gio.max()/gio.min()-1:.0%}")
print()
print(f"TONG SO VAN DE: {loi}")
````

## A5 — Rà báo cáo: câu nào nêu số trung bình mà thiếu phân rã

Cell dưới quét `docs/bao_cao_tuan3.md`, tìm mọi câu nêu chỉ số tổng thể (MAE, MAPE, coverage,
độ rộng) rồi kiểm xem **quanh đó có nhắc tới phân rã không** (band, theo giờ, theo nhóm…).

Câu bị đánh dấu ⚠️ là chỗ mentor sẽ đọc và hỏi *"trung bình của cái gì?"*. **Không sửa báo cáo
tuần 3** (đã nộp) — dùng danh sách này khi viết tuần 4.

````python
import re

BC = Path("docs/bao_cao_tuan3.md")
dong = BC.read_text(encoding="utf-8").split("\n")

# Cau nao neu chi so tong the
CHI_SO = re.compile(r"MAE|MAPE|coverage|độ rộng|R²|±\s*\d")
# Dau hieu da co phan ra o gan do
PHAN_RA = re.compile(r"band|phân rã|theo giờ|theo nhóm|từng|mỗi|điều kiện|Mondrian|"
                     r"quãng đường|<50k|>300k|cao điểm", re.I)

canh_bao, on = [], 0
for i, d in enumerate(dong):
    if not CHI_SO.search(d) or d.strip().startswith("|---"):
        continue
    quanh = "\n".join(dong[max(0, i-4): i+5])
    if PHAN_RA.search(quanh):
        on += 1
    else:
        canh_bao.append((i+1, d.strip()))

print(f"Quet {len(dong):,} dong cua {BC.name}")
print(f"  Cau neu chi so tong the: {len(canh_bao)+on}")
print(f"  Da co phan ra gan do   : {on}")
print(f"  THIEU phan ra          : {len(canh_bao)}")
print()
print("=" * 78)
print("DANH SACH CAN BO SUNG PHAN RA KHI VIET BAO CAO TUAN 4")
print("=" * 78)
for ln, d in canh_bao:
    print(f"  dong {ln:>4}: {d[:110]}")
print()
print("Goi y bo sung cho tung loai:")
print("  - Cau ve coverage  -> them 'nhung band >300k chi 82,7%' (uncertainty/04_SO_SANH)")
print("  - Cau ve MAE/MAPE  -> them 'deu qua moi khung gio, chenh 0,003 diem' (uncertainty/05_PHAN_RA_theo_thoi_gian)")
print("  - Cau ve do rong   -> them so theo band (hinh BM2)")
````
